```python
from pathlib import Path
import pandas as pd

project = Path.cwd()

if not (project / "dataset").exists():
    project = project.parent

dataset = project / "dataset"

print("Week 10: LangChain Basics")
print("Project folder:", project)
print("Dataset folder exists:", dataset.exists())
print("CSV files:", [p.name for p in dataset.glob("*.csv")])
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

llm = ChatOllama(model="llama3.2")

# Original prompt
original = PromptTemplate.from_template(
    "Answer this question: {question}"
)

# Improved prompt
improved = PromptTemplate.from_template(
    """You are a helpful College Assistant.
    Answer clearly using simple language.
    Give practical steps when appropriate.
    Never invent official college information.
    Question: {question}"""
)

question = "How can I prepare for semester exams?"

print("ORIGINAL PROMPT OUTPUT:")
print((original | llm).invoke({"question": question}).content)

print("\nIMPROVED PROMPT OUTPUT:")
print((improved | llm).invoke({"question": question}).content)
```

```python
# Chain 1: Summarization
summary_prompt = PromptTemplate.from_template(
    "Summarize this text in 3 bullet points: {text}"
)
summary_chain = summary_prompt | llm

print(summary_chain.invoke({
    "text": "Students should plan their study time, review notes, "
            "practice questions, take breaks, and sleep well."
}).content)

# Chain 2: Question answering
qa_prompt = PromptTemplate.from_template(
    "Context: {context}\nQuestion: {question}\nAnswer:"
)
qa_chain = qa_prompt | llm

print(qa_chain.invoke({
    "context": "The college library opens at 9 AM on weekdays.",
    "question": "When does the library open on weekdays?"
}).content)
```
